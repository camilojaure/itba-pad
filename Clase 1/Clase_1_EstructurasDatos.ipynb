## 1. Variables y tipos de datos

Una variable es un nombre que le ponemos a un valor para poder reutilizarlo.  
Python tiene cuatro tipos básicos de datos:

```python
# str → texto (string)
nombre    = "Ana García"
provincia = "Buenos Aires"
segmento  = "Premium"

# int → número entero
edad              = 34
cant_productos    = 3
antiguedad_anios  = 7

# float → número con decimales
balance     = 250000.75
tasa_anual  = 0.085

# bool → verdadero o falso
cliente_activo = True
tiene_churn    = False

print(type(nombre))
print(type(edad))
print(type(balance))
print(type(cliente_activo))
```

**¿Por qué importa el tipo?** Porque define qué podés hacer con el valor.

```python
# Con números podés operar
print(f'Balance con interés anual: ${balance * (1 + tasa_anual):,.2f}')
print(f'Edad en meses: {edad * 12}')

# Con texto podés manipular
print(nombre.upper())
print(nombre.split())
print(f'{nombre} vive en {provincia}, segmento {segmento}')
```

```python
# bool en condiciones
if cliente_activo:
    print(f'{nombre} es un cliente activo.')
else:
    print(f'{nombre} no está activo.')

# bool surge de comparaciones
print(balance > 100000)   # True
print(edad < 18)          # False
print(segmento == 'Joven') # False
```

## 2. Listas

Una lista es una secuencia ordenada de valores. Es el antecedente directo de una columna en un DataFrame de pandas.

```python
# Crear una lista
balances   = [920000, 42000, 18000, 215000, 875000, 12000, 38000]
segmentos  = ['Premium', 'Retail', 'Joven', 'PyME', 'Premium', 'Joven', 'Retail']
churn_list = [0, 1, 1, 0, 0, 1, 0]

print(balances)
print(f'Cantidad de clientes: {len(balances)}')
```

```python
# Indexación — los índices arrancan en 0
print(f'Primer cliente:  ${balances[0]:,}')
print(f'Último cliente:  ${balances[-1]:,}')
print(f'Primeros tres:   {balances[:3]}')
print(f'Últimos dos:     {balances[-2:]}')
```

```python
# Métodos de lista
balances.append(760000)    # agregar al final
print(f'Con nuevo cliente: {balances}')
print(f'Balance máximo:    ${max(balances):,}')
print(f'Balance mínimo:    ${min(balances):,}')
print(f'Balance total:     ${sum(balances):,}')
print(f'Balance promedio:  ${sum(balances)/len(balances):,.0f}')
```

```python
# Iterar sobre una lista — recorrer todos los elementos
for i, (seg, bal) in enumerate(zip(segmentos, balances[:len(segmentos)])):
    print(f'Cliente {i+1}: {seg:8s} → ${bal:>10,}')
```

```python
# List comprehension — filtrar en una línea
balances_altos = [b for b in balances if b > 100000]
print(f'Clientes con balance > $100k: {balances_altos}')
```

## 3. Diccionarios

Un diccionario almacena pares **clave: valor**. Es la forma más natural de representar un registro (una fila de datos).

```python
# Un cliente como diccionario
cliente = {
    'nombre':         'Ana García',
    'edad':           34,
    'segmento':       'Premium',
    'balance_ars':    250000.75,
    'cliente_activo': True,
    'churn':          0
}

print(cliente)
print(type(cliente))
```

```python
# Acceder a valores
print(cliente['nombre'])
print(cliente['balance_ars'])

# Agregar o modificar
cliente['provincia'] = 'CABA'
cliente['balance_ars'] = 275000.0

# Ver todas las claves
print(list(cliente.keys()))
print(list(cliente.values()))
```

```python
# Una lista de diccionarios → el paso previo a un DataFrame
clientes = [
    {'nombre': 'Ana García',    'segmento': 'Premium', 'balance': 250000, 'churn': 0},
    {'nombre': 'Luis Romero',   'segmento': 'Joven',   'balance': 12000,  'churn': 1},
    {'nombre': 'María Sosa',    'segmento': 'Retail',  'balance': 38000,  'churn': 0},
    {'nombre': 'Carlos Pérez',  'segmento': 'PyME',    'balance': 215000, 'churn': 0},
    {'nombre': 'Sofía Torres',  'segmento': 'Joven',   'balance': 9000,   'churn': 1},
]

# Iterar
for c in clientes:
    estado = 'se fue' if c['churn'] == 1 else 'activo'
    print(f"{c['nombre']:20s} | {c['segmento']:8s} | ${c['balance']:>10,} | {estado}")
```

## 4. De Python a pandas

Acá es donde todo se conecta. pandas toma estas estructuras y las convierte en DataFrames — las tablas con las que vamos a trabajar durante todo el curso.

```python
import pandas as pd

# De lista de diccionarios → DataFrame
df = pd.DataFrame(clientes)
df
```

```python
# De diccionario de listas → DataFrame (otra forma)
df2 = pd.DataFrame({
    'segmento': segmentos,
    'balance':  balances[:len(segmentos)],
    'churn':    churn_list
})
df2
```

**La conexión:**

| Estructura Python | Equivalente en pandas |
|---|---|
| Lista | Una columna (Series) |
| Lista de diccionarios | DataFrame completo |
| Diccionario de listas | DataFrame completo |
| Tipo `str` | Columna `object` |
| Tipo `int` / `float` | Columna `int64` / `float64` |
| Tipo `bool` | Columna `bool` |

```python
# Una vez en pandas, todo lo que vimos en Python sigue valiendo
print(f"Balance promedio: ${df['balance'].mean():,.0f}")
print(f"Tasa de churn:    {df['churn'].mean():.0%}")
print()

# Y aparecen métodos mucho más poderosos
df.groupby('segmento')['balance'].mean().sort_values(ascending=False)
```

## 5. Objetos y métodos: la lógica detrás de la notación

Todo en Python es un objeto. La notación `objeto.método()` aparece en todas partes.

```python
# El objeto define qué métodos están disponibles

texto = "ana garcía"
print(texto.upper())       # str tiene .upper()
print(texto.title())       # str tiene .title()
print(texto.replace('a', 'A'))  # str tiene .replace()

lista = [5, 2, 8, 1, 9, 3]
lista.sort()               # list tiene .sort()
print(lista)
lista.reverse()            # list tiene .reverse()
print(lista)
```

```python
# Un DataFrame de pandas también es un objeto — con muchos más métodos
print(type(df))            # <class 'pandas.core.frame.DataFrame'>

# Algunos métodos que vamos a usar durante todo el curso:
# df.head()        → primeras filas
# df.info()        → tipos y nulos
# df.describe()    → estadísticas
# df.groupby()     → agrupar
# df.merge()       → unir tablas
# df.fillna()      → rellenar nulos
# df.drop()        → eliminar columnas o filas

print('Métodos disponibles en un DataFrame:', len([m for m in dir(df) if not m.startswith('_')]))
```

## 6. Qué es una librería

Una librería es un conjunto de código que alguien ya escribió y empaquetó para que otros lo reutilicen.

```python
# Cada import trae herramientas nuevas
import pandas as pd          # tablas y manipulación de datos
import matplotlib.pyplot as plt  # gráficos
import seaborn as sns         # gráficos con menos código

# Las librerías también son objetos — con sus propias funciones y clases
print(type(pd))        # <module 'pandas'>
print(type(plt))       # <module 'matplotlib.pyplot'>
```

```python
# La misma tarea con y sin librería

# SIN pandas: calcular el promedio de balance manualmente
total = sum(balances)
promedio_manual = total / len(balances)
print(f'Promedio manual:  ${promedio_manual:,.0f}')

# CON pandas: una línea
serie = pd.Series(balances)
print(f'Promedio pandas:  ${serie.mean():,.0f}')
print(f'Mediana pandas:   ${serie.median():,.0f}')
print(f'Desvío pandas:    ${serie.std():,.0f}')
```

## Resumen

| Concepto | Para qué sirve | En pandas |
|---|---|---|
| `str` | Texto: nombres, categorías, provincias | Columnas `object` |
| `int` / `float` | Números: balance, edad, tasa | Columnas numéricas |
| `bool` | Verdadero/falso: activo, churn | Columnas booleanas / filtros |
| Lista | Secuencia de valores del mismo tipo | Una columna (Series) |
| Diccionario | Par clave:valor — un registro | Una fila |
| Objeto | Todo en Python | DataFrame, Series, Plot |
| Método | Acción que ejecuta un objeto | `.mean()`, `.groupby()`, `.head()` |
| Librería | Herramientas pre-escritas | `pandas`, `matplotlib`, `seaborn` |
