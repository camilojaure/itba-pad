## Antes de arrancar: dos columnas que conviene tener claras

Varias preguntas de abajo se apoyan en estas dos, y no significan lo mismo.

**`churn`** — vale `1` si el cliente se fue, `0` si sigue. En esta base, *irse* significa
que **dio de baja sus productos o dejó de operar por completo durante los últimos doce meses**.

Ojo con esto: no es una verdad universal, es **la definición que tomó el equipo que armó los
datos**. Con otra definición —por ejemplo *"sin actividad en los últimos 90 días"*— los mismos
clientes darían otra tasa de churn. Por eso, cuando alguien te muestre una tasa, la primera
pregunta nunca es cuánto da: es cómo la definió.

**`cliente_activo`** — vale `True` si el cliente **registra actividad reciente**: movimientos,
consumos, uso de la cuenta. Es una foto del presente, no del cierre de la relación.

Como miden cosas distintas, las dos columnas **no coinciden fila por fila**:

- casi todos los que fugaron figuran inactivos, que es lo esperable
- pero hay clientes **inactivos que no están marcados como churn**: no dieron de baja nada,
  simplemente dejaron de usar la cuenta

Ese segundo grupo es el caso ambiguo que discutimos en la teoría. Con una definición de churn
por inactividad pasarían a contar como fuga y la tasa de la cartera subiría bastante. Con la
definición de esta base, no cuentan. Ninguna de las dos está mal — miden fenómenos distintos,
y cada una habilita decisiones distintas.
