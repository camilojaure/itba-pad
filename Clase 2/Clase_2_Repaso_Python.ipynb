## 1. Una variable es un nombre con un valor adentro


````python
nombre = "Ana García"
edad = 34

print(nombre)
print(edad)
````

Cada valor pertenece a una **clase**. La clase es lo que define qué se puede hacer
con ese valor. Se la podés preguntar con `type()`:


````python
print(type(nombre))
print(type(edad))
````

Mirá bien lo que contestó: **`<class 'str'>`**. Le preguntaste el *tipo* y te respondió
con la palabra **clase**. No es un error de Python — en Python **tipo y clase son la misma cosa**,
y de acá en adelante nosotros vamos a decir siempre **clase**.

Hay cuatro clases básicas. Con datos de un cliente:

| Clase | Qué guarda | Ejemplo |
|---|---|---|
| `str` | texto | `"Ana García"`, `"Premium"` |
| `int` | número entero | `34`, `3` |
| `float` | número con decimales | `250000.75` |
| `bool` | verdadero o falso | `True`, `False` |


````python
segmento = "Premium"      # str
balance  = 250000.75     # float
activo   = True          # bool

print(type(segmento))
print(type(balance))
print(type(activo))
````

### Pará un segundo antes de correr esto

La clase define qué podés hacer con el valor. Mirá las dos celdas de abajo
y **predecí qué va a imprimir cada una** antes de ejecutarlas.


````python
# Con números, el + suma
print(edad + 1)
````

````python
# Con texto, el + pega uno atrás del otro
print(nombre + " es cliente Premium")
````

Esa es toda la diferencia. Mismo símbolo, dos comportamientos, según la clase.

## 2. Una lista es una columna

Si una variable guarda un valor, una lista guarda muchos. Es lo más parecido a
una columna de Excel.


````python
balances  = [250000, 89000, 430000, 12000]
segmentos = ["Premium", "Joven", "Retail", "PyME"]

print(balances)
print(len(balances))
````

## 3. Python es un lenguaje orientado a objetos

Es la definición que van a leer en cualquier lado, y suena mucho más difícil de lo que es.
La forma más rápida de entenderla es compararla con Excel.

**En Excel, el dato es pasivo.** La celda `A1` guarda `"ana garcía"` y no sabe hacer nada.
Si querés pasarla a mayúsculas, traés una fórmula *de afuera* y le pasás la celda como argumento:

```
=MAYUSC(A1)
```

**En Python, el dato viene con sus acciones adentro.** El texto `"ana garcía"` ya sabe ponerse
en mayúsculas. No traés nada de afuera: se lo pedís al dato directamente, con un punto:

```
nombre.upper()
```

> Esa es toda la idea de *orientado a objetos*: en vez de datos por un lado y funciones por otro,
> cada dato viene empaquetado junto con las cosas que sabe hacer.
> La fórmula ya no rodea al dato — el dato la trae puesta.


### La jerarquía: clase → objeto → método

Son tres palabras y un solo concepto. Van en este orden, y cada una vive adentro de la anterior:

```
CLASE      str                      el molde
  └── OBJETO   "Ana García"         un ejemplar hecho con ese molde
        └── MÉTODO   .upper()       una acción que el molde le dio
```

**1 · La clase es el molde.** No es un dato: es la definición de qué puede guardar y qué puede
hacer. `str` es la clase de todos los textos que existen y van a existir.

**2 · El objeto es el ejemplar concreto.** `"Ana García"` es *un* texto hecho con ese molde.
`"Premium"` es otro. Al ejemplar también se lo llama **instancia** — instancia de `str`.
Molde hay uno; ejemplares, millones.

**3 · El método es cada acción que el molde le puso adentro.** Como `.upper()` está definido en
la clase `str`, **todos** los objetos `str` saben hacerlo. Ninguno tiene que aprenderlo por su cuenta.

> **La analogía de Excel:** la clase es el **formato de celda**, el objeto es **la celda**, y el
> método es **lo que ese formato te habilita a hacer**. Dos celdas con `10/03/2026` adentro se ven
> igual; si una está formateada como fecha y la otra como texto, una te deja restarle otra fecha
> y la otra no. El contenido no cambió nada — cambió el molde, y el molde define el menú.

Y esto escala sin cambiar de idea: más adelante `DataFrame` va a ser la clase, `df` el objeto,
y `.head()` uno de sus métodos. Misma jerarquía, otro molde.


### El método, en detalle

Un **método** se llama con un punto:

```
objeto.metodo()
```

Tres cosas para leer esa línea:

- **`objeto`** → a quién se lo estás pidiendo
- **`.`** → "pedile a este objeto que…"
- **`metodo()`** → qué le pedís. **Los paréntesis significan "ejecutalo ahora"**


````python
# nombre es un objeto de la clase str. Estos son dos de sus métodos.
print(nombre.upper())
print(nombre.lower())
````

Fijate que `nombre` no cambió: `.upper()` **devuelve** un texto nuevo en mayúsculas.
Otros métodos sí modifican el objeto sobre el que se llaman. `append`, en las listas, es uno:


````python
balances.append(75000)   # agrega un valor al final de la lista

print(balances)
````

### Cada clase trae su propio menú

Si la clase es el molde, entonces **un objeto solo puede hacer lo que su clase le dio**.
`.upper()` está definido en `str`. En la clase `int`, no.

#### Pará un segundo antes de correr esto

`edad` guarda `34`, un objeto de la clase `int`. ¿Qué pasa si le pedimos que se ponga en mayúsculas?


````python
edad.upper()
````

Da error, y el error dice exactamente lo que pasa:

```
AttributeError: 'int' object has no attribute 'upper'
```

*"El objeto de clase `int` no tiene ningún `upper`."* No está roto: le pediste al objeto algo que
su molde nunca le dio. Es el equivalente a pedirle `=MAYUSC()` a una celda con un número adentro.

> **Los errores de Python se leen, no se esquivan.** Casi siempre te dicen la clase del objeto y el
> nombre de lo que no encontró. Con esas dos cosas ya sabés dónde mirar.

**Truco para la clase:** escribí `nombre.` (con el punto) y apretá la tecla `Tab`.
Colab te despliega el menú completo de métodos que la clase `str` le dio a ese objeto.


### La equivalencia con Excel, ordenada

| Lo que querés hacer | Excel | Python |
|---|---|---|
| Promedio | `=PROMEDIO(A:A)` | `.mean()` |
| Contar | `=CONTAR(A:A)` | `.count()` |
| Máximo | `=MAX(A:A)` | `.max()` |
| Mayúsculas | `=MAYUSC(A1)` | `.upper()` |

Los nombres cambian, pero lo que cambia de verdad es el **orden**: en Excel la fórmula va primero
y el dato entra adentro; en Python el dato va primero y la acción va pegada atrás.

> Si te acordás de una sola cosa de esta sección, que sea leer `df.head()` como
> *"che df, mostrame la cabeza"*. Todo el curso es esa frase, con distintos verbos.

## 5. De listas sueltas a una tabla

Acá se conecta todo: pandas toma las listas y arma con ellas una tabla, que es
con lo que vamos a trabajar el resto del curso. Se llama **DataFrame**.
Cada lista es una columna; cada fila es un cliente.


````python
df = pd.DataFrame({
    "segmento": ["Premium", "Joven", "Retail", "PyME"],
    "balance":  [250000, 12000, 38000, 215000],
    "churn":    [0, 1, 0, 0]
})

df
````

Y una vez que es una tabla, los métodos siguen funcionando igual — pero ahora
sobre una columna entera:


````python
print(df["balance"].mean())
print(df["churn"].mean())
````

> Ese último número es la **tasa de churn**: la proporción de clientes que se fueron.
> Con cuatro clientes no dice nada. Dentro de un rato lo vamos a calcular sobre
> mil clientes reales, y ahí sí vamos a poder decidir algo.


### El DataFrame también es un objeto — el más importante del curso

Acá se cierra la jerarquía de la sección 3, con el molde que nos importa de verdad:

```
CLASE      DataFrame        el molde, lo trajo pandas
  └── OBJETO   df           nuestra tabla, hecha con ese molde
        └── MÉTODO   .head()   una acción que el molde le dio
```

`DataFrame` no es una clase que trae Python: **la trajo pandas** cuando hicimos el `import`.
Eso es lo que hace una librería — te da moldes nuevos que no venían de fábrica. Comprobalo:


````python
print(type(df))
````

Ahora bien: un objeto grande como este trae dos cosas distintas, y conviene separarlas desde hoy.


````python
print(df.shape)    # ATRIBUTO: sin paréntesis
print(df.head(2))  # MÉTODO: con paréntesis
````

| | Qué es | Cómo se escribe |
|---|---|---|
| **Atributo** | Un dato que el objeto **ya tiene**. No calcula nada, te lo muestra | `df.shape` |
| **Método** | Una acción que el objeto **ejecuta** cuando se lo pedís | `df.head(2)` |

La regla práctica: **los paréntesis son "hacelo ahora"**. Si no hay paréntesis, no se ejecuta nada,
solo se consulta algo que ya estaba ahí.

> En Excel tenés la misma distinción sin nombrarla: que un rango tenga 4 filas y 3 columnas es un
> **dato del rango** — lo leés abajo a la derecha, no lo calculás. `=PROMEDIO()` en cambio corre
> en el momento y devuelve un resultado nuevo. Atributo y método.

**Error clásico de la primera semana:** escribir `df.shape()` con paréntesis. Python responde
`TypeError: 'tuple' object is not callable` — le pediste que ejecutara algo que no es una acción.


Y una última cosa, que va a aparecer todo el tiempo: **un método puede devolver otro objeto,
que a su vez tiene sus propios métodos.** Por eso se ven líneas con varios puntos encadenados:


````python
# value_counts() devuelve una tabla nueva, y a esa tabla le pedimos head(2)
df["segmento"].value_counts().head(2)
````

Se lee de izquierda a derecha, como una frase: *de `df`, la columna `segmento`, contá cuántos
hay de cada uno, y de eso mostrame los dos primeros.*

> No hay magia nueva acá: es `objeto.metodo()` tres veces seguidas. Cada resultado es un objeto
> que trae su propio menú.
